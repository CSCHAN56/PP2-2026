15. 두 정수의 함과 차를 각각 반환하는 함수를 완성하시오.
def add_sub(a,b)
	result1 = a + b
	result2 = a – b
	return (result1,result2)
x,y = add_sub(7,2)
print(x,y)
함수 a,b에 7,2를 받고 result1,2의 값을 구하는 식을 만든 다음 return (result1,result2)으로 함수 호출한 변수 x,y의 값을 반환하면 완료
