#counter 클래스 만들기
-데이터: 현재 숫자
-동작: 증가하기, 현재 값 읽어오기

```python
class Counter:
    def __init__(self, initial_count=0):
        self.count = initial_count

    def increament(self):
        self.count += 1

    def decrement(self):
        self.count -= 1

    def reset(self):
        self.count = 0

    def get_count(self):
        return self.count

def test_counter():
    myc = Counter()
    print(myc.get_count())
    myc.increament()
    print(myc.get_count())

    yourc = Counter(5)
    print(yourc.get_count())
    yourc.increament()
    print(yourc.get_count())
    yourc.increament()
    print(yourc.get_count())

test_counter()
    
```

```python
class Student :
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}")

kim = Student("Kim", 21)
kim.display_info()
print(kim.name)
```
