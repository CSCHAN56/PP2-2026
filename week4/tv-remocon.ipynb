#TV 클래스 추가
- 데이터: 현재 채널번호,볼륨업,on여부
- 동작: 켜기,끄기,채널변경,볼륨변경,현재 TV 객체 내용 출력

- 클래스명: mytelevision

```python
class mytelevision:
    def __init__(self, channel=1, volume=5, is_on=False):
        self.channel = channel
        self.volume = volume
        self.is_on = is_on

    def turn_on(self):
        self.is_on = True

    def turn_off(self):
        self.is_on = False

    def change_channel(self, channel):
        self.channel = channel

    def change_volume(self, volume):
        self.volume = volume

    def display_status(self):
        print(f"Channel: {self.channel}, Volume: {self.volume}, Is On: {self.is_on}")

def test_television():
    mytv = mytelevision(channel = 11)
    mytv.turn_on()
    mytv.display_status()
    mytv.turn_off()
    mytv.display_status()

test_television()
```
