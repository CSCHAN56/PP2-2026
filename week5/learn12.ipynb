```python
# 일반적인 자동차를 나타내는 클래스이다.
class Car:
    def __init__(self, maker, model, color, price):
        self.maker = maker # 메이커
        self.model = model # 모델
        self.color = color # 자동차의 색상
        self.price = price # 자동차의 가격

    def setDesc(self, maker): # 설정자 메소드
        self.maker = maker

    # 수정 1: getDesc가 중복 선언되어 있었으며, 이 부분은 getMaker여야 합니다.
    def getMaker(self): # 접근자 메소드 
        return self.maker
    
    def getDesc(self): # 접근자 메소드
        return self.__str__()

    def __str__(self): # 객체를 문자열로 변환하는 메소드
        return f"Maker: {self.maker}, Model: {self.model}, Color: {self.color}, Price: {self.price}"

class ElectricCar(Car): # 전기차를 나타내는 클래스이다.
    def __init__(self, maker, model, color, price, bat_size):
        super().__init__(maker, model, color, price)
        self.batterysize = bat_size # 배터리 크기

    def setBatterSize(self, bat_size): # 설정자 메소드
        self.batterysize = bat_size

    def getBatterSize(self): # 접근자 메소드
        return self.batterysize

    def __str__(self):
        return f"{super().__str__()}, Battery : {self.batterysize}"

    def getDesc(self): # 접근자 메소드
        return self.__str__()

# 수정 2: Truck 클래스의 들여쓰기를 바깥으로 빼주어야 합니다.
class Truck(Car):
    def __init__(self, maker, model, color, price, payLoad): 
        super().__init__(maker, model, color, price)
        self.payLoad = payLoad # 적재량

    def __str__(self):
        return f"{super().__str__()}, PayLoad : {self.payLoad}kg"

# --- 실행 부분 ---
car1 = Car("Kia", "Ray", "Blue", 2000)
print(car1.getDesc())

ecar1 = ElectricCar("Tesla", "Model S", "Red", 10000, 81)
print(ecar1)

ecar1.setBatterSize(92)
print(ecar1.getDesc())

truck1 = Truck("Hyundai", "Porter", "White", 4000, 1000)
print(truck1)
print(truck1.getMaker())
```
