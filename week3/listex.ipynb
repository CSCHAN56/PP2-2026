LIST연습 노트 파일입니다.

# 인덱스(index)란 리스트에서의 항목의 위(번호)이다.(0부터 시작한다.)
# insert(),append()는 리스트에 추가하는 명령어이다.
# pop(),remove()는 항목의 저장된 리스트를 삭제한다.
# "value" in mylist 는 "value"가 mylist에 있다면 True를 반환한다.
# "value" not in mylist 는 "value"가 mylist에 있지 않다면 True를 반환한다.
# mylist.sort()는 mylist를 정렬하는 연산이다.
# mylist.index("value")는 "value"가 발견된 위치를 반환하는 연산이다.
# mylist.append("value")는 리스트 끝에 "value"요소를 추가하는 연산이다.
# mylist.remove("value")는 mylist에서 "value"가 나타나는 위치를 찾아서 삭제하는 연산이다.
