- lab) GUI코드

```python
from tkinter import *      #tkinter 모듈생성

window = Tk()
window.title("Widgets Test")       #창이름
window.geometry("400x800")         #창크기

left_frame = Frame(window, bg="gray")  #왼쪽 프레임
right_frame = Frame(window, bg="gray") #오른쪽 프레임

label1 = Label(left_frame, text = "Button:",bg="yellow")     #버튼 라벨1
label2 = Label(left_frame, text = "TextField:",bg="yellow")  #엔트리 라벨2

button = Button(left_frame, text = "Button",  #버튼
        bg = "white", fg = "black",
        width = 10, height = 2)

entry = Entry(left_frame, width = 10)   #엔트리

right_frame.pack(side="right", fill="both", expand=True)
left_frame.pack(side="left", fill="both", expand=True)
label1.grid(row=0, column=0)
button.grid(row=0, column=1)
label2.grid(row=1, column=0)
entry.grid(row=1, column=1)
window.mainloop()
```
